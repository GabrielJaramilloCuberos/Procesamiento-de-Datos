#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a, b):
    """
    Compara dos números: si ambos son pares, devuelve el menor.
    Si al menos uno es impar, devuelve el mayor.
    """
    result = 0  # Inicializa la variable que guardará el resultado

    # Verifica si AMBOS números (a y b) son pares
    if a % 2 == 0 and b % 2 == 0:
        # Si ambos son pares, busca el menor de los dos
        if a < b:
            result = a
        else:
            result = b
    # Ejecuta este bloque si al menos uno de los números es impar
    else:
        # Si hay un impar, busca el mayor de los dos
        if a > b:
            result = a
        else:
            result = b

    return result  # Devuelve el valor final obtenido
```

```python
# Revisión
lesser_of_two_evens(2,4)
```

```python
# Revisión
lesser_of_two_evens(2,5)
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
    """
    Verifica si las dos primeras palabras de una cadena de texto
    comienzan con la misma letra. Devuelve True o False.
    """
    # Divide el texto en una lista de palabras usando los espacios como separador
    palabras = text.split()

    # Compara la primera letra de la primera palabra (palabras[0][0])
    # con la primera letra de la segunda palabra (palabras[1][0])
    if palabras[0][0] == palabras[1][0]:
        return True   # Devuelve True si las letras iniciales coinciden
    else:
        return False  # Devuelve False si las letras iniciales son diferentes
```

```python
# Revisión
animal_crackers('Ciudad Calor')
```

```python
# Revisión
animal_crackers('Lugar Frio')
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1, n2):
    """
    Verifica si alguno de los dos números es 20,
    o si la suma de ambos da exactamente 20.
    """
    # Evalúa tres condiciones con el operador lógico 'or' (O):
    # 1. Si el primer número es igual a 20
    # 2. Si el segundo número es igual a 20
    # 3. Si la suma de ambos números es igual a 20
    if n1 == 20 or n2 == 20 or n1 + n2 == 20:
        return True   # Devuelve True si al menos una condición se cumple
    else:
        return False  # Devuelve False si ninguna de las condiciones es verdadera
```

```python
# Revisión
makes_twenty(20,10)
```

```python
# Revisión
makes_twenty(2,3)
```

```python
makes_twenty(12,8)
```

#### MAYÚSCULAS: Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto
     
    old_macdonald('macdonald') --> MacDonald
    
Note: `'macdonald'.capitalize()` returns `'Macdonald'`

```python
def old_macdonald(name):
    """
    Capitaliza la primera y la cuarta letra de un nombre.
    Ejemplo: 'macdonald' se convierte en 'MacDonald'.
    """
    # name[:3] toma los primeros 3 caracteres (índices 0, 1, 2) y .capitalize() hace mayúscula la primera letra.
    # name[3:] toma el resto de la cadena (desde el índice 3 en adelante) y .capitalize() hace mayúscula su primera letra (la cuarta del nombre original).
    # El operador + une ambas partes transformadas en un solo texto.
    return name[:3].capitalize() + name[3:].capitalize()
```

```python
# Check
old_macdonald('macdonald')
```

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    """
    Invierte el orden de las palabras en una cadena de texto.
    Ejemplo: 'Home I am' se convierte en 'am I Home'.
    """
    # Divide el texto en una lista de palabras individuales usando los espacios como separador
    palabras = text.split()

    # Invierte el orden de los elementos dentro de la lista 'palabras' (modifica la lista original)
    palabras.reverse()

    # Une las palabras de la lista invertida en una sola cadena de texto,
    # separando cada elemento con un espacio en blanco
    return " ".join(palabras)
```

#### PROBLEMA 33:

Dada una lista de enteros, retornar True si el arreglo tinene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python
def has_33(nums):
    """
    Verifica si en una lista de números existen dos números 3 seguidos.
    Devuelve True si los encuentra, de lo contrario devuelve False.
    """
    # Recorre la lista desde el índice 0 hasta el penúltimo elemento (len(nums)-1).
    # Se detiene en el penúltimo para evitar que nums[i+1] busque fuera de la lista.
    for i in range(0, len(nums)-1):

        # Compara si el número actual Y el número siguiente son iguales a 3
        if nums[i] == 3 and nums[i+1] == 3:
            return True  # Termina la función de inmediato y devuelve True si los encuentra

    # Si el bucle termina y nunca encontró la pareja de 3, devuelve False
    return False
```

```python
# Revisión
has_33([1, 3, 3])
```

```python
# Revisión
has_33([1, 3, 1, 3])
```

```python
# Revisión
has_33([3, 1, 3])
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):
    """
    Toma una cadena de texto y devuelve otra donde cada
    carácter original se repite exactamente tres veces.
    Ejemplo: 'Hello' se convierte en 'HHHeeellllllooo'.
    """
    resultado = ''  # Inicializa una cadena vacía para ir guardando el texto final

    # Recorre la cadena original carácter por carácter
    for char in text:
        # Multiplica el carácter actual por 3 (ej. 'a' * 3 = 'aaa')
        # y lo suma (concatena) al final de la variable 'resultado'
        resultado += char * 3

    return resultado  # Devuelve la nueva cadena con los caracteres triplicados
```

```python
# Check
paper_doll('Hello')
```

```python
# Check
paper_doll('Mississippi')
```
